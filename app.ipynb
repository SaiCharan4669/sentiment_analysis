```python
import gradio as gr
import joblib

# Load model and vectorizer
model = joblib.load("sentiment_tfidf_ngrams.pkl")
tfidf = joblib.load("tfidf_vectorizer.pkl")

def predict_sentiment(review):

    if not review.strip():
        return "Please enter a review!", "", ""

    # Transform review
    transformed = tfidf.transform([review])

    # Prediction
    prediction = model.predict(transformed)[0]

    # Probabilities
    proba = model.predict_proba(transformed)[0]

    neg_prob = f"{proba[0]*100:.2f}%"
    pos_prob = f"{proba[1]*100:.2f}%"

    confidence = max(proba) * 100

    if prediction == "positive":
        result = f" Positive Sentiment ({confidence:.2f}% confidence)"
    else:
        result = f" Negative Sentiment ({confidence:.2f}% confidence)"

    return result, pos_prob, neg_prob


examples = [
    ["This movie was absolutely amazing, best film I have seen in years!"],
    ["Terrible movie, complete waste of time and money."],
    ["Brilliant direction, outstanding performances."],
    ["The plot made no sense and the acting was awful."]
]

with gr.Blocks() as app:

    gr.Markdown("#  Movie Reviews Sentiment Analyzer")

    review_input = gr.Textbox(
        label="Enter Movie Review",
        lines=5,
        placeholder="Type your review here..."
    )

    predict_btn = gr.Button("Analyze")

    result = gr.Textbox(label="Prediction")
    pos_prob = gr.Textbox(label="Positive Probability")
    neg_prob = gr.Textbox(label="Negative Probability")

    gr.Examples(
        examples=examples,
        inputs=review_input
    )

    predict_btn.click(
        fn=predict_sentiment,
        inputs=review_input,
        outputs=[result, pos_prob, neg_prob]
    )

app.launch()
```
